## 7) Análisis de complejidad **[5 pts]**

Estrategia: Se utiliza programación dinámica con memorización. El programa guarda los resultados que ya calculó para no repetirlos.

Función T(n): Como se pueden calcular hasta n valores y tenemos 5 tipos de monedas:
T(n) = c₁ + n · 5 · c₂

Complejidad temporal: O(n)
Esto se debe a que el programa puede calcular diferentes valores desde 0 hasta el monto ingresado y para cada valor revisa las 5 monedas.

Complejidad espacial: O(n)
Se utiliza memoria para guardar los resultados en memo.

Mejor caso: O(1)
Si el monto es 0, el programa termina inmediatamente.

Peor caso: O(n)
Si el monto es grande y todavía no hay resultados guardados, se deben calcular y almacenar muchos valores.

## 8) Código  funcional **[10 pts]**
Incluye tu implementación completa.

No pegues código incompleto; debe ser ejecutable.


```python
monedas = [50, 100, 200, 500, 1000]
memo = {}

def min_monedas(monto):

    if monto == 0:
        return 0

    if monto in memo:
        return memo[monto]

    mejor = monto + 1

    for moneda in monedas:
        if monto >= moneda:
            cantidad = min_monedas(monto - moneda) + 1

            if cantidad < mejor:
                mejor = cantidad

    memo[monto] = mejor
    return mejor


monto = int(input("Ingrese el valor del cambio: "))

while monto < 0 or monto % 50 != 0:
    print("No es posible dar el cambio exacto, ingrese de nuevo")
    monto= int (input())
else:
    resultado = min_monedas(monto)
    print("Cantidad mínima de monedas:", resultado)
```

#### Código base para benchmarking

```python
# Calcula el tiempo de ejecucion con base a diferentes tamaños de la entrada
# y nos da una lista de resultados, cada resultado contiene el tamaño de la entrada
# y el tiempo de ejecucion
# TENER EN CUENTA ESTAS ENTRADAS PUEDEN VARIAR SEGÚN EL PROBLEMA


def bench(n: int, step: int = 1) -> list[tuple[int, float]]:
    print("Generando datos ...", end='')
    from random import randint
    import time
    results: list[tuple[int, float]] = []


    # AJUSTAR LAS ENTRADAS SEGÚN EL PROBLEMA
    for size in range(1, n, step):
        input = size

        ti = time.perf_counter()
        output = input # Ajustar a su función
        tf = time.perf_counter()


        delta = (tf - ti) * 1000
        print(".", end="")
        results.append((size, delta))
    return results


# === CODIGO PARA DIBUJAR ===


def dibujar(x, y, dimensiones=(7, 5)):
    assert len(x) == len(y)
    import matplotlib.pyplot as plt
    plt.figure(figsize=dimensiones)
    plt.plot(x, y)
    plt.show()


# === LLAMADO ===
tiempos = bench(10, step=1)


x = [size for size, timing in tiempos]
y = [timing for size, timing in tiempos]


dibujar(x, y)
```
