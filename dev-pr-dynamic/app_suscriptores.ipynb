## 1) Descripción del problema **[1 pts]**


Una plataforma de streaming esta analizando el crecimiento de nuevos suscriptores por mes.

Se conoce que
 - En el mes 0 se registraron 0 usuarios nuevos
 - En el mes 1 se registro 1 usuario nuevo
 - A partir del mes 2, el numero de usuarios es igual a la suma de los nuevos usuarios de los dos meses anteriores, debido al efecto combinado de referidos del mes anterior y recomendaciones acumuladas del mes previo

 La idea es determinar con memorizacion cuantos nuevos suscriptores habra en el mes n

## 6) Código funcional **[15 pts]**


```python
memo={}
def suscriptores(n):
  if n==0:
    return 0
  if n==1:
    return 1
  if n in memo:
    return memo[n]
  memo[n]= suscriptores(n-1) + suscriptores(n-2)
  return memo[n]

n=int(input("ingrese el mes que desea saber: "))

if n<0:
  print("Error: el mes no puede ser negativo")
else:
  resultado=suscriptores(n)
  print("En el mes",n,"tendras",resultado,"nuevos suscriptores")
```
